```python
import time
import random
from dataclasses import dataclass

# -------------------------------
# Food Profile (Static Properties)
# -------------------------------
@dataclass
class FoodProfile:
    name: str
    ideal_temp: float           # °C
    required_energy: float      # Joules
    water_content: float        # %
    max_safe_temp: float        # °C
    softness_threshold: float   # %

# -------------------------------
# Heat Sensor Simulation
# -------------------------------
class HeatSensor:
    def read_temperature(self):
        # Simulated temperature reading (replace with real sensor)
        return random.uniform(60, 95)

# -------------------------------
# Cooking Controller
# -------------------------------
class SmartCooker:
    def __init__(self, food: FoodProfile, mass_kg: float):
        self.food = food
        self.mass = mass_kg
        self.c = 4186  # J/kg°C (water-based food)
        self.sensor = HeatSensor()

        self.prev_temp = 25.0
        self.energy_accumulated = 0.0
        self.time_elapsed = 0

    # -------------------------------
    # Energy Calculation (Physics)
    # -------------------------------
    def calculate_energy(self, current_temp):
        delta_t = max(current_temp - self.prev_temp, 0)
        energy = self.mass * self.c * delta_t
        self.prev_temp = current_temp
        return energy

    # -------------------------------
    # Softness Calculation
    # -------------------------------
    def calculate_softness(self, temp):
        energy_ratio = min(self.energy_accumulated / self.food.required_energy, 1)
        temp_ratio = min(temp / self.food.ideal_temp, 1)
        return (energy_ratio * temp_ratio) * 100

    # -------------------------------
    # Safety Check
    # -------------------------------
    def safety_check(self, temp):
        if temp > self.food.max_safe_temp:
            print("⚠️  Safety Alert: Temperature too high! Reducing heat.")
            return False
        return True

    # -------------------------------
    # Main Cooking Loop
    # -------------------------------
    def cook(self):
        print(f"\n🍳 Cooking Started: {self.food.name}")
        print("-" * 40)

        while True:
            temp = self.sensor.read_temperature()
            energy = self.calculate_energy(temp)
            self.energy_accumulated += energy
            self.time_elapsed += 5

            softness = self.calculate_softness(temp)

            print(
                f"Time: {self.time_elapsed:>3}s | "
                f"Temp: {temp:.2f}°C | "
                f"Energy: {self.energy_accumulated:.0f} J | "
                f"Softness: {softness:.2f}%"
            )

            if not self.safety_check(temp):
                break

            if softness >= self.food.softness_threshold:
                print("\n✅ Food Cooked Successfully!")
                print(f"🟢 Final Softness: {softness:.2f}%")
                print(f"⏱ Total Time: {self.time_elapsed} seconds")
                break

            time.sleep(1)  # simulate 5-second sensor interval

# -------------------------------
# Example Run
# -------------------------------
if __name__ == "__main__":
    chicken = FoodProfile(
        name="Chicken",
        ideal_temp=75,
        required_energy=100000,
        water_content=65,
        max_safe_temp=95,
        softness_threshold=90
    )

    cooker = SmartCooker(food=chicken, mass_kg=0.5)
    cooker.cook()
```

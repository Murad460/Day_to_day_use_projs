```python
# Algorithm Surface_Tension_Physics_Model

# Input:
#     F        // force (N)
#     L        // length (m)
#     T        // temperature
#     T0       // reference temperature
#     k        // temperature coefficient
#     rho      // density of liquid
#     r        // capillary radius
#     theta    // contact angle

# Constants:
#     g = 9.81

# Process:

# 1. gamma = F / L

# 2. gamma_T = gamma * (1 - k * (T - T0))

# 3. h = (2 * gamma_T * cos(theta)) / (rho * g * r)

# Output:
#     gamma_T
#     h
```

```python
import math

class SurfaceTensionModel:
    def __init__(self, g=9.81):
        self.g = g  # gravity (m/s^2)

    # 1. Base surface tension
    def surface_tension(self, force, length):
        """
        γ = F / L
        """
        return force / length

    # 2. Temperature correction
    def temperature_corrected_tension(self, gamma_0, T, T0, k):
        """
        γ(T) = γ0 * (1 - k(T - T0))
        """
        return gamma_0 * (1 - k * (T - T0))

    # 3. Capillary rise calculation
    def capillary_rise(self, gamma, density, radius, theta_deg):
        """
        h = (2 * γ * cosθ) / (ρ g r)
        """
        theta_rad = math.radians(theta_deg)
        h = (2 * gamma * math.cos(theta_rad)) / (density * self.g * radius)
        return h


# -------------------------------
# Example Usage
# -------------------------------
if __name__ == "__main__":

    model = SurfaceTensionModel()

    # Input values
    force = 0.072        # Newton
    length = 1.0         # meter
    T = 30               # current temperature (°C)
    T0 = 20              # reference temperature (°C)
    k = 0.00015          # temperature coefficient
    density = 1000       # kg/m^3 (water)
    radius = 0.0005      # meters
    theta = 0            # degrees (water-glass)

    # Calculations
    gamma_0 = model.surface_tension(force, length)
    gamma_T = model.temperature_corrected_tension(gamma_0, T, T0, k)
    height = model.capillary_rise(gamma_T, density, radius, theta)

    # Output
    print("Base Surface Tension (γ0):", gamma_0, "N/m")
    print("Temperature Corrected γ:", gamma_T, "N/m")
    print("Capillary Rise Height:", height, "meters")
```

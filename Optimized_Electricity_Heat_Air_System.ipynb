```python
# Algorithm Optimized_Electricity_Heat_Air_System

# Input:
#     V      // Voltage
#     I      // Current
#     t      // Time
#     R      // Heater resistance
#     m      // Mass of heated material
#     c      // Specific heat capacity
#     CFM    // Airflow rate
#     A      // Fan area
#     Pmotor // Motor power
#     Rloss  // Internal loss resistance

# Constants:
#     rho = 1.225   // air density

# Compute:
#     Ein = V * I * t

#     Eheat = I^2 * R * t
#     deltaT = Eheat / (m * c)

#     v = (CFM * 0.0004719) / A
#     Eair = 0.5 * rho * A * v^3 * t

#     Eloss = Ein - (Eheat + Eair)

#     heat_eff = (Eheat / Ein) * 100
#     airflow_eff = (Eair / (Pmotor * t)) * 100
#     total_eff = ((Eheat + Eair) / Ein) * 100

#     R_optimal = Rloss
#     max_efficiency = R / (R + Rloss)

# Output:
#     deltaT
#     airflow_eff
#     total_eff
#     R_optimal
#     max_efficiency
```

```python
import math

class EnergyConversionSystem:
    def __init__(
        self,
        voltage,          # Volts
        current,          # Amperes
        time_sec,         # Seconds
        heater_resistance,# Ohms
        mass,             # kg (heated object / air)
        specific_heat,    # J/kg°C
        cfm,              # Cubic feet per minute
        fan_area,         # m²
        motor_power,      # Watts
        loss_resistance   # Ohms (internal/system loss)
    ):
        self.V = voltage
        self.I = current
        self.t = time_sec
        self.R = heater_resistance
        self.m = mass
        self.c = specific_heat
        self.CFM = cfm
        self.A = fan_area
        self.P_motor = motor_power
        self.R_loss = loss_resistance

        self.rho = 1.225  # air density kg/m³

    # ---------------- ENERGY ---------------- #

    def electrical_energy_input(self):
        return self.V * self.I * self.t

    def heat_energy(self):
        return (self.I ** 2) * self.R * self.t

    def air_velocity(self):
        # Convert CFM to m³/s then to velocity
        return (self.CFM * 0.0004719) / self.A

    def air_energy(self):
        v = self.air_velocity()
        return 0.5 * self.rho * self.A * (v ** 3) * self.t

    def energy_loss(self):
        return self.electrical_energy_input() - (
            self.heat_energy() + self.air_energy()
        )

    # ---------------- TEMPERATURE ---------------- #

    def temperature_rise(self):
        return self.heat_energy() / (self.m * self.c)

    # ---------------- EFFICIENCY ---------------- #

    def heat_efficiency(self):
        return (self.heat_energy() / self.electrical_energy_input()) * 100

    def airflow_efficiency(self):
        return (self.air_energy() / (self.P_motor * self.t)) * 100

    def total_efficiency(self):
        return (
            (self.heat_energy() + self.air_energy())
            / self.electrical_energy_input()
        ) * 100

    # ---------------- OPTIMIZATION ---------------- #

    def optimal_resistance(self):
        # Maximum Power Transfer Theorem
        return self.R_loss

    def max_possible_efficiency(self):
        return self.R / (self.R + self.R_loss)

    # ---------------- REPORT ---------------- #

    def report(self):
        print("\n--- ENERGY CONVERSION REPORT ---\n")
        print(f"Input Energy (J): {self.electrical_energy_input():.2f}")
        print(f"Heat Energy (J): {self.heat_energy():.2f}")
        print(f"Air Energy (J): {self.air_energy():.2f}")
        print(f"Energy Loss (J): {self.energy_loss():.2f}\n")

        print(f"Temperature Rise (°C): {self.temperature_rise():.2f}\n")

        print(f"Heat Efficiency (%): {self.heat_efficiency():.2f}")
        print(f"Airflow Efficiency (%): {self.airflow_efficiency():.2f}")
        print(f"Total Efficiency (%): {self.total_efficiency():.2f}\n")

        print(f"Optimal Resistance (Ohms): {self.optimal_resistance():.2f}")
        print(f"Max Possible Efficiency: {self.max_possible_efficiency() * 100:.2f}%")
        print("\n--------------------------------\n")


# ---------------- EXAMPLE USAGE ---------------- #

if __name__ == "__main__":
    system = EnergyConversionSystem(
        voltage=230,            # V
        current=5,              # A
        time_sec=60,            # seconds
        heater_resistance=10,   # ohms
        mass=1.2,               # kg
        specific_heat=1005,     # J/kg°C (air)
        cfm=150,                # airflow
        fan_area=0.05,          # m²
        motor_power=200,        # watts
        loss_resistance=5       # ohms
    )

    system.report()
```
